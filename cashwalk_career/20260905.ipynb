## 세팅 (먼저 실행)

````python
import time

TOTAL_SECONDS = 70 * 60
_start = None

def start_timer():
    """70분 카운트다운을 시작합니다."""
    global _start
    _start = time.time()
    print("타이머 시작. 70:00 남음. 행운을 빕니다.")

def time_left():
    """남은 시간을 출력합니다."""
    if _start is None:
        print("아직 start_timer() 를 실행하지 않았습니다.")
        return
    remaining = TOTAL_SECONDS - (time.time() - _start)
    if remaining <= 0:
        print("시간 종료. 지금까지 푼 것으로 채점하세요.")
        return
    m, s = divmod(int(remaining), 60)
    print(f"남은 시간 {m:02d}:{s:02d}")


def _normalize(x):
    """리스트/튜플 중첩 구조를 리스트로 통일해서 비교합니다."""
    if isinstance(x, (list, tuple)):
        return [_normalize(i) for i in x]
    return x


def check(fn, cases, name=""):
    """cases: [(args_tuple, expected), ...]. CodeSignal처럼 통과 개수를 보여줍니다."""
    passed = 0
    for i, (args, expected) in enumerate(cases, 1):
        try:
            got = fn(*args)
            ok = _normalize(got) == _normalize(expected)
        except Exception as e:
            print(f"  Test {i}: 예외 발생 -> {type(e).__name__}: {e}")
            continue
        if ok:
            passed += 1
            print(f"  Test {i}: PASS")
        else:
            print(f"  Test {i}: FAIL")
            print(f"     input    = {args}")
            print(f"     expected = {expected}")
            print(f"     got      = {got}")
    total = len(cases)
    tail = "  << 전부 통과!" if passed == total else ""
    print(f"\n{name} 결과: {passed}/{total} 통과{tail}")
    return passed, total
````

````python
start_timer()
````

## Q1 — Above Average

Given an array of integers `nums`, return the sum of all elements that are **strictly greater**
than the average of the array. If the array is empty, return `0`.

**Example**

```
nums = [1, 5, 3, 9, 2]
average = 4.0
elements above average: 5, 9
return 14
```

**Constraints**

- `0 <= len(nums) <= 10^5`
- `-10^9 <= nums[i] <= 10^9`

````python
def solution_q1(nums: list[int]) -> int:
    if not nums:
        return 0

    avg = sum(nums) / len(nums)

    total = 0
    for v in nums:
        if v > avg:
            total += v

    return total
````

````python
cases_q1 = [
    (([1, 5, 3, 9, 2],), 14),
    (([],), 0),
    (([5, 5, 5],), 0),
    (([-10, 0, 10],), 10),
    (([1],), 0),
    (([-5, -1],), -1),
    (([1000000000, -1000000000, 0],), 1000000000),
    (([3, 1, 4, 1, 5, 9, 2, 6],), 24),
]
check(solution_q1, cases_q1, "Q1")
````
